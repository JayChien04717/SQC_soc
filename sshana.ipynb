```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from scipy.special import erf
from scipy.integrate import quad
from copy import deepcopy
from itertools import cycle
import scipy as sp
# ===================================================================== #
# Section 1: Math & Physics Helper Functions (數學與物理模型)
# ===================================================================== #


# ====================================================== #
# Gaussian Fit Functions
# ====================================================== #
def gaussian(x, a, x0, sigma, y0):
    return a * np.exp(-((x - x0) ** 2) / (2 * sigma**2)) + y0


def fit_gauss(xdata, ydata, fitparams=None):
    # xmed, xstd should be gotten from the single shot data prior to fitting the histogram
    if fitparams is None:
        fitparams = [None] * 4
    else:
        fitparams = np.copy(fitparams)
    if fitparams[0] is None:
        fitparams[0] = np.max(ydata)
    if fitparams[1] is None:
        fitparams[1] = xdata[np.argmax(ydata)]
    if fitparams[2] is None:
        fitparams[2] = (np.max(xdata) - np.min(xdata)) / 8
    if fitparams[3] is None:
        fitparams[3] = np.min(ydata)
    pOpt = fitparams
    # print('fitparams guess:', fitparams)
    pCov = np.full(shape=(len(fitparams), len(fitparams)), fill_value=np.inf)
    bounds = (
        [fitparams[0] * 0.5, np.min(xdata), fitparams[2] * 0.1, 0],
        [fitparams[0] * 1.5, np.max(xdata), fitparams[2] * 10, np.max(ydata)],
    )
    for i, param in enumerate(fitparams):
        if not (bounds[0][i] < param < bounds[1][i]):
            fitparams[i] = np.mean((bounds[0][i], bounds[1][i]))
            print(
                f"Attempted to init fitparam {i} to {param}, which is out of bounds {bounds[0][i]} to {bounds[1][i]}. Instead init to {fitparams[i]}"
            )
    try:
        pOpt, pCov = sp.optimize.curve_fit(
            gaussian,
            xdata,
            ydata,
            p0=np.array(fitparams, dtype="float64"),
            bounds=bounds,
        )
        # return pOpt, pCov
    except RuntimeError:
        print("Warning: fit failed!")
        # return 0, 0
    return pOpt, pCov


# ====================================================== #
# Double Gaussian Fit Functions
# ====================================================== #


def double_gaussian(x, a1, b1, c1, a2, b2, c2):
    """
    Standard double Gaussian function.
    Params: a=amplitude, b=mean, c=sigma
    """
    return a1 * np.exp(-((x - b1) ** 2) / (2 * c1**2)) + a2 * np.exp(
        -((x - b2) ** 2) / (2 * c2**2)
    )


def fit_doublegauss(xdata, ydata, fitparams):
    """
    Robust fitting function for double Gaussian distributions.

    Args:
        xdata: Bin centers.
        ydata: Counts/Histogram values.
        fitparams: Initial guesses [amp_g, mu_g, sigma_g, amp_e, mu_e, sigma_e].

    Returns:
        pOpt: Optimized parameters.
        pCov: Covariance matrix.
    """
    # Unpack parameters to set dynamic bounds
    ag, mug, sig, ae, mue, sie = fitparams

    # --- Constraint Logic ---
    # Prevent the solver from drifting too far from the expected state centers (mu).
    # If the signal for one state is weak, the solver might try to merge both Gaussians
    # into the single visible peak. We constrain the mean to +/- 2 sigma (or a fixed buffer).

    delta_mu_g = abs(sig) * 2 if sig > 0 else 100
    delta_mu_e = abs(sie) * 2 if sie > 0 else 100

    # Lower Bounds: Amp > 0, Sigma > 0, Mean constrained
    lb = [0, mug - delta_mu_g, 0, 0, mue - delta_mu_e, 0]
    # Upper Bounds: Amp infinite, Sigma infinite, Mean constrained
    ub = [np.inf, mug + delta_mu_g, np.inf, np.inf, mue + delta_mu_e, np.inf]

    try:
        pOpt, pCov = sp.optimize.curve_fit(
            double_gaussian,
            xdata,
            ydata,
            p0=fitparams,
            bounds=(lb, ub),
            maxfev=10000,  # Increase max iterations for convergence
        )
    except Exception as e:
        # print(f"Fit failed: {e}")
        # If fit fails, return the initial guesses so the code doesn't crash
        pOpt = fitparams
        pCov = np.zeros((6, 6))

    return pOpt, pCov


def rotate(x, y, theta):
    """Rotate points in the x-y plane by angle theta."""
    return (
        x * np.cos(theta) - y * np.sin(theta),
        x * np.sin(theta) + y * np.cos(theta),
    )


def full_rotate(d, theta):
    """Rotate all IQ data in a dictionary by angle theta."""
    d = deepcopy(d)
    d["Ig"], d["Qg"] = rotate(d["Ig"], d["Qg"], theta)
    d["Ie"], d["Qe"] = rotate(d["Ie"], d["Qe"], theta)
    if "If" in d and "Qf" in d:
        d["If"], d["Qf"] = rotate(d["If"], d["Qf"], theta)
    return d


def make_hist(d, nbins=200):
    """Create a normalized histogram from data (Helper for calculations)."""
    hist_vals, bin_edges = np.histogram(d, bins=nbins, density=True)
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
    return bin_centers, hist_vals


def gaussian(x, mag, cen, wid):
    """Basic Gaussian function."""
    return mag / (np.sqrt(2 * np.pi) * wid) * np.exp(-((x - cen) ** 2) / (2 * wid**2))


def double_gaussian(x, mag1, cen1, wid1, mag2, cen2, wid2):
    """Sum of two Gaussians."""
    return gaussian(x, mag1, cen1, wid1) + gaussian(x, mag2, cen2, wid2)


def fit_gauss(x, y, p0):
    """Wrapper for scipy curve_fit with gaussian."""
    return curve_fit(lambda x, mag, cen, wid: gaussian(x, mag, cen, wid), x, y, p0=p0)


def fit_doublegauss(x, y, p0):
    """Wrapper for scipy curve_fit with double gaussian."""
    return curve_fit(double_gaussian, x, y, p0=p0)


# --- T1 Decay Physics Models ---


def distfn(v, vg, ve, sigma, tm):
    """
    Distribution function modeling T1 decay during measurement.
    tm = t_meas / T1
    """
    dv = ve - vg
    # Avoid division by zero
    if dv == 0:
        dv = 1e-9

    term1 = tm * sigma**2 / (2 * dv**2)
    term2 = (v - vg) / dv

    # Arguments for error functions
    arg1 = (tm * sigma**2 / dv + ve - v) / (np.sqrt(2) * sigma)
    arg2 = (-tm * sigma**2 / dv + v - vg) / (np.sqrt(2) * sigma)

    return np.abs(
        (tm / (2 * dv)) * np.exp(tm * (term1 - term2)) * (erf(arg1) + erf(arg2))
    )


def excited_func(x, vg, ve, sigma, tm):
    """
    Fit function for excited state including T1 decay effects.
    Combines Gaussian (remaining e) + T1 decay tail.
    """
    # Note: Assuming normalized magnitude of 1 for the total distribution
    return gaussian(x, 1, ve, sigma) * np.exp(-tm) + distfn(x, vg, ve, sigma, tm)


def fit_gaussian_helper(d, nbins=100, p0=None, plot=False):
    """Helper to fit a simple gaussian to data array."""
    v, hist_val = make_hist(d, nbins)
    if p0 is None:
        p0 = [np.mean(v * hist_val) / np.mean(hist_val), np.std(d)]

    try:
        # Fit Gaussian with fixed magnitude of 1 (normalized)
        params, _ = curve_fit(
            lambda x, cen, wid: gaussian(x, 1, cen, wid), v, hist_val, p0=p0
        )
    except RuntimeError:
        params = p0

    return params, v, hist_val


# ===================================================================== #
# Section 2: Core Visualization (通用繪圖邏輯)
# ===================================================================== #

default_colors = plt.rcParams["axes.prop_cycle"].by_key()["color"]
linestyle_cycle = ["solid", "dashed", "dotted", "dashdot"]
marker_cycle = ["o", "*", "s", "^"]


def plot_hist_line(
    data, bins, ax, xlims=None, color=None, linestyle=None, label=None, alpha=None
):
    """Helper to plot histogram as a line graph."""
    hist_data, bin_edges = np.histogram(data, bins=bins, range=xlims)
    # Normalize
    if hist_data.sum() > 0:
        hist_data = hist_data / hist_data.sum()

    for i in range(len(hist_data)):
        lbl = label if i == 0 else None
        ax.plot(
            [bin_edges[i], bin_edges[i + 1]],
            [hist_data[i], hist_data[i]],
            color=color,
            linestyle=linestyle,
            label=lbl,
            alpha=alpha,
            linewidth=0.9,
        )
        if i < len(hist_data) - 1:
            ax.plot(
                [bin_edges[i + 1], bin_edges[i + 1]],
                [hist_data[i], hist_data[i + 1]],
                color=color,
                linestyle=linestyle,
                alpha=alpha,
                linewidth=0.9,
            )
    ax.relim()
    ax.set_ylim((0, None))
    return hist_data, bin_edges


def general_hist(
    iqshots,
    state_labels,
    g_states,
    e_states,
    e_label="e",
    check_qubit_label=None,
    numbins=200,
    amplitude_mode=False,
    ps_threshold=None,
    theta=None,
    plot=True,
    verbose=True,
    fid_avg=False,
    fit=False,
    gauss_overlap=False,
    plotoverlap=False,
    fitparams=None,
    normalize=True,
    title=None,
    export=False,
):
    """
    Main visualization engine.
    Processes raw IQ shots, calculates fidelity, plots histograms and confusion matrix.
    """
    if numbins is None:
        numbins = 200
    has_f_state = len(iqshots) > 2

    # --- 1. Data Aggregation ---
    data_map = {"g": np.array([]), "e": np.array([]), "f": np.array([])}
    I_tot_all = np.array([])
    Q_tot_all = np.array([])

    for check_i, data_check in enumerate(iqshots):
        I, Q = data_check
        I_tot_all = np.concatenate((I_tot_all, I))
        Q_tot_all = np.concatenate((Q_tot_all, Q))

        if check_i in g_states:
            cat = "g"
        elif check_i in e_states:
            cat = "e"
        else:
            cat = "f"

        if data_map[cat].size == 0:
            data_map[cat] = I + 1j * Q
        else:
            data_map[cat] = np.concatenate((data_map[cat], I + 1j * Q))

    # --- 2. Rotation Calculation ---
    if not amplitude_mode:
        if theta is None:
            xg = np.mean(np.real(data_map["g"])) if data_map["g"].size > 0 else 0
            yg = np.mean(np.imag(data_map["g"])) if data_map["g"].size > 0 else 0
            xe = np.mean(np.real(data_map["e"])) if data_map["e"].size > 0 else 1
            ye = np.mean(np.imag(data_map["e"])) if data_map["e"].size > 0 else 1
            theta = -np.arctan2((ye - yg), (xe - xg))
        else:
            theta *= np.pi / 180

        def rotate_iq_internal(c_data, ang):
            return np.real(c_data) * np.cos(ang) - np.imag(c_data) * np.sin(ang)

        I_all_new = rotate_iq_internal(I_tot_all + 1j * Q_tot_all, theta)
        span = (np.max(I_all_new) - np.min(I_all_new)) / 2
        midpoint = (np.max(I_all_new) + np.min(I_all_new)) / 2
    else:
        theta = 0
        amp_all = np.abs(I_tot_all + 1j * Q_tot_all)
        span = (np.max(amp_all) - np.min(amp_all)) / 2
        midpoint = (np.max(amp_all) + np.min(amp_all)) / 2

    xlims = [midpoint - span, midpoint + span]

    # --- 3. Plot Setup ---
    if plot:
        fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(9, 7))
        if title is None:
            title = f"Readout Fidelity" + (
                f" on Q{check_qubit_label}" if check_qubit_label is not None else ""
            )
        fig.suptitle(title)
        fig.tight_layout()
        plt.subplots_adjust(hspace=0.35, wspace=0.15)

        # Subplot settings
        axs[0, 0].set_title("Unrotated")
        axs[0, 0].axis("equal")
        axs[0, 1].set_title(f"Rotated ($\\theta={theta * 180 / np.pi:.1f}^\\circ$)")
        axs[0, 1].axis("equal")
        threshold_axis = "I" if not amplitude_mode else "Amplitude"
        axs[1, 0].set_xlabel(f"{threshold_axis} [ADC levels]")
        axs[1, 0].set_ylabel("Counts")

    # Variables
    n_dist = {"g": None, "e": None, "f": None}
    bins_dist = None
    gauss_fit_fidelity = 0
    popts = []
    pcovs = []
    b_g_plot, c_g_plot = None, None
    b_e_plot, c_e_plot = None, None

    # --- 4. Process Each Input State ---
    for check_i, data_check in enumerate(iqshots):
        state_label = state_labels[check_i]
        I, Q = data_check
        complex_data = I + 1j * Q

        this_color = default_colors[check_i % len(default_colors)]

        # Rotation
        if not amplitude_mode:
            I_new = rotate_iq_internal(complex_data, theta)
            Q_new = np.real(complex_data) * np.sin(theta) + np.imag(
                complex_data
            ) * np.cos(theta)
            data_to_hist = I_new
        else:
            I_new, Q_new = I, Q  # Placeholder
            data_to_hist = np.abs(complex_data)

        # Plot Scatter
        if plot:
            axs[0, 0].scatter(
                I,
                Q,
                label=state_label,
                color=this_color,
                marker=".",
                alpha=0.1,
                edgecolor="None",
            )
            axs[0, 0].plot(
                [np.mean(I)],
                [np.mean(Q)],
                color="k",
                marker="o",
                markerfacecolor=this_color,
            )
            axs[0, 1].scatter(
                I_new,
                Q_new,
                label=state_label,
                color=this_color,
                marker=".",
                alpha=0.1,
                edgecolor="None",
            )

        # Plot/Calc Histogram
        if plot:
            n, bins = plot_hist_line(
                data_to_hist,
                bins=numbins,
                ax=axs[1, 0],
                xlims=xlims,
                color=this_color,
                label=state_label,
                alpha=0.6,
            )
        else:
            n, bins = np.histogram(data_to_hist, bins=numbins, range=xlims)

        bins_dist = bins

        # Accumulate
        if check_i in g_states:
            cat = "g"
        elif check_i in e_states:
            cat = "e"
        else:
            cat = "f"

        if n_dist[cat] is None:
            n_dist[cat] = n
        else:
            n_dist[cat] += n

    # --- 5. Fitting Logic ---
    # ... (Reuse your Gaussian overlap logic here, simplified for brevity) ...
    # This part remains mostly the same as your provided code
    if (
        (fit or gauss_overlap or plotoverlap)
        and n_dist["g"] is not None
        and n_dist["e"] is not None
    ):
        # [INSERT GAUSSIAN FITTING CODE HERE FROM YOUR ORIGINAL SNIPPET]
        # Assuming fit_doublegauss or fit_gauss are called here
        pass

    # --- 6. Thresholds & Fidelity ---
    fids = []
    thresholds = []

    # G-E Threshold
    contrast_ge = np.abs(
        (np.cumsum(n_dist["g"]) - np.cumsum(n_dist["e"]))
        / (np.sum(n_dist["g"]) + np.sum(n_dist["e"]))
    )
    tind_ge = contrast_ge.argmax()
    threshold_ge = bins_dist[tind_ge]
    thresholds.append(threshold_ge)
    fids.append(contrast_ge[tind_ge])

    # --- 7. Matrix & Final Plotting ---
    if not has_f_state:
        matrix_size = 2
        labels = ["|g>", f"|{e_label}>"]
        n00 = n_dist["g"][:tind_ge].sum()
        n01 = n_dist["g"][tind_ge:].sum()
        n10 = n_dist["e"][:tind_ge].sum()
        n11 = n_dist["e"][tind_ge:].sum()
        raw_matrix = np.array([[n00, n01], [n10, n11]])
    else:
        # Simple 3-state thresholding logic
        matrix_size = 3
        labels = ["|g>", f"|{e_label}>", "|f>"]
        if n_dist["f"] is not None:
            # E-F Threshold
            contrast_ef = np.abs(
                (np.cumsum(n_dist["e"]) - np.cumsum(n_dist["f"]))
                / (np.sum(n_dist["e"]) + np.sum(n_dist["f"]))
            )
            tind_ef = contrast_ef.argmax()
            thresholds.append(bins_dist[tind_ef])

            ts = sorted([tind_ge, tind_ef])

            def classify(n):
                return [n[: ts[0]].sum(), n[ts[0] : ts[1]].sum(), n[ts[1] :].sum()]

            raw_matrix = np.array(
                [classify(n_dist["g"]), classify(n_dist["e"]), classify(n_dist["f"])]
            )
        else:
            raw_matrix = np.zeros((3, 3))

    row_sums = raw_matrix.sum(axis=1)[:, np.newaxis]
    row_sums[row_sums == 0] = 1
    conf_matrix = 100 * raw_matrix / row_sums

    if plot:
        # Draw thresholds and matrix
        for th in thresholds:
            axs[1, 0].axvline(th, color="k", linestyle="--")
        axs[1, 0].set_title(f"Fidelity: {100 * fids[0]:.2f}%")
        axs[1, 0].legend()

        ax_cm = axs[1, 1]
        im = ax_cm.imshow(conf_matrix, cmap="Reds", vmin=0, vmax=100)
        ax_cm.set_xticks(range(matrix_size))
        ax_cm.set_yticks(range(matrix_size))
        ax_cm.set_xticklabels(labels)
        ax_cm.set_yticklabels(labels)
        for i in range(matrix_size):
            for j in range(matrix_size):
                ax_cm.text(
                    j,
                    i,
                    f"{conf_matrix[i, j]:.1f}",
                    ha="center",
                    va="center",
                    color="white" if conf_matrix[i, j] > 50 else "black",
                )

        if export:
            plt.savefig("readout.jpg", dpi=600)
            plt.close()
        else:
            plt.show()

    # --- 8. Data Packaging for Advanced Fit ---
    # [CRITICAL STEP] Pack data for fit_single_shot
    sorted_data_dict = {
        "Ig": np.real(data_map["g"]),
        "Qg": np.imag(data_map["g"]),
        "Ie": np.real(data_map["e"]),
        "Qe": np.imag(data_map["e"]),
    }
    if has_f_state:
        sorted_data_dict["If"] = np.real(data_map["f"])
        sorted_data_dict["Qf"] = np.imag(data_map["f"])

    # Returns
    if gauss_overlap:
        return_data = [gauss_fit_fidelity, thresholds, theta * 180 / np.pi]
    else:
        return_data = [fids, thresholds, theta * 180 / np.pi]

    if fit or gauss_overlap or plotoverlap:
        return_data += [popts, pcovs]

    # Append the dictionary at the end
    return_data.append(sorted_data_dict)

    if verbose:
        print(f"Theta: {theta * 180 / np.pi:.2f} deg")
        print(f"Fidelity: {100 * fids[0]:.3f}%")
        print("Matrix:\n", conf_matrix)

    return return_data


# ===================================================================== #
# Section 3: Advanced T1 Analysis (進階物理分析)
# ===================================================================== #


def fit_single_shot(d, plot=True, rot=True):
    """
    Comprehensive analysis of single-shot readout data with T1 decay model.
    Accepts the dictionary output from general_hist().
    """
    d = deepcopy(d)

    # 1. Rotation (If requested)
    if rot:
        # Use simple geometry to find optimal rotation
        xg, yg = np.mean(d["Ig"]), np.mean(d["Qg"])
        xe, ye = np.mean(d["Ie"]), np.mean(d["Qe"])
        theta = -np.arctan2((ye - yg), (xe - xg))
        data = full_rotate(d, theta)
    else:
        data = d
        theta = 0

    # 2. Fit Gaussians to G (I & Q) and E (Q only)
    # We use these to establish baseline noise (sigma) and centers
    paramsg, vhg, histg = fit_gaussian_helper(data["Ig"], nbins=100)
    paramsqg, vhqg, histqg = fit_gaussian_helper(data["Qg"], nbins=100)
    paramsqe, vhqe, histqe = fit_gaussian_helper(data["Qe"], nbins=100)

    # Extract parameters
    vg = paramsg[0]  # I center for ground
    sigma = paramsg[1]  # Width (from ground state)
    vqg = paramsqg[0]  # Q center for ground
    vqe = paramsqe[0]  # Q center for excited

    # 3. Fit E (I axis) with T1 Decay Model
    vhe, histe = make_hist(data["Ie"], nbins=100)

    # Initial guess: [center_of_e, decay_rate_tm]
    p0 = [np.mean(vhe * histe) / np.mean(histe), 0.2]

    # Constrain fit: Fix vg and sigma from ground state fit
    try:
        paramse, _ = curve_fit(
            lambda x, ve, tm: excited_func(x, vg, ve, sigma, tm), vhe, histe, p0=p0
        )
    except RuntimeError:
        print("T1 Fit failed, using guess.")
        paramse = p0

    ve = paramse[0]
    tm = paramse[1]  # t_meas / T1
    paramse2 = [vg, ve, sigma, tm]

    # 4. Plotting
    if plot:
        plt.figure(figsize=(8, 5))
        plt.plot(vhg, histg, "b.", alpha=0.3, label="Data G")
        plt.plot(vhe, histe, "r.", alpha=0.3, label="Data E")

        # Plot Fits
        x_dense = np.linspace(min(vhg[0], vhe[0]), max(vhg[-1], vhe[-1]), 500)
        plt.plot(x_dense, gaussian(x_dense, 1, vg, sigma), "b-", lw=2, label="Fit G")
        plt.plot(
            x_dense,
            excited_func(x_dense, vg, ve, sigma, tm),
            "r-",
            lw=2,
            label=f"Fit E (Tm={tm:.2f})",
        )

        plt.ylabel("Probability")
        plt.xlabel("I [Rotated]")
        plt.title(f"T1 Decay Analysis ($T_m/T_1 = {tm:.3f}$)")
        plt.legend()
        plt.tight_layout()
        plt.show()

    # 5. Return
    fit_results = {
        "theta": theta,
        "vg": vg,
        "ve": ve,
        "sigma": sigma,
        "tm": tm,
        "vqg": vqg,
        "vqe": vqe,
    }
    return data, fit_results


# ===================================================================== #
# Section 4: Wrappers (為了相容性)
# ===================================================================== #


def hist(data, **kwargs):
    """Wrapper for dictionary input to general_hist"""
    iqshots = [(data["Ig"], data["Qg"]), (data["Ie"], data["Qe"])]
    state_labels = ["g", "e"]
    g_states = kwargs.pop("g_states", [0])
    e_states = kwargs.pop("e_states", [1])

    if "If" in data:
        iqshots.append((data["If"], data["Qf"]))
        state_labels.append("f")

    return general_hist(
        iqshots=iqshots,
        state_labels=state_labels,
        g_states=g_states,
        e_states=e_states,
        **kwargs,
    )
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from scipy.special import erf
from scipy.integrate import quad
from copy import deepcopy
from itertools import cycle
import scipy as sp

# ===================================================================== #
# Section 1: Math & Physics Helper Functions
# ===================================================================== #


def rotate(x, y, theta):
    """Rotate points in the x-y plane by angle theta."""
    return (
        x * np.cos(theta) - y * np.sin(theta),
        x * np.sin(theta) + y * np.cos(theta),
    )


def full_rotate(d, theta):
    """Rotate all IQ data in a dictionary by angle theta."""
    d = deepcopy(d)
    d["Ig"], d["Qg"] = rotate(d["Ig"], d["Qg"], theta)
    d["Ie"], d["Qe"] = rotate(d["Ie"], d["Qe"], theta)
    if "If" in d and "Qf" in d:
        d["If"], d["Qf"] = rotate(d["If"], d["Qf"], theta)
    return d


def make_hist(d, nbins=200):
    """Create a normalized histogram from data."""
    hist_vals, bin_edges = np.histogram(d, bins=nbins, density=True)
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
    return bin_centers, hist_vals


# --- Core Fitting Models ---


def gaussian(x, mag, cen, wid):
    """
    Normalized Gaussian (PDF).
    Note: We removed y0 (offset) to ensure physical probability interpretation.
    """
    return mag / (np.sqrt(2 * np.pi) * wid) * np.exp(-((x - cen) ** 2) / (2 * wid**2))


def double_gaussian(x, mag1, cen1, wid1, mag2, cen2, wid2):
    """Sum of two Gaussians."""
    return gaussian(x, mag1, cen1, wid1) + gaussian(x, mag2, cen2, wid2)


def fit_gauss_helper(x, y, p0=None):
    """Wrapper for fitting single gaussian."""
    try:
        popt, _ = curve_fit(
            lambda x, mag, cen, wid: gaussian(x, mag, cen, wid), x, y, p0=p0
        )
    except RuntimeError:
        popt = p0
    return popt


def fit_doublegauss_helper(xdata, ydata, fitparams):
    """
    Robust double Gaussian fit with bounds.
    fitparams: [mag1, cen1, wid1, mag2, cen2, wid2]
    """
    # Unpack for readability
    a1, b1, c1, a2, b2, c2 = fitparams

    # Dynamic Bounds to prevent fit form diverging
    # Allow centers to drift slightly, but not swap or drift too far
    delta_b1 = abs(c1) * 3 if c1 > 0 else 1.0
    delta_b2 = abs(c2) * 3 if c2 > 0 else 1.0

    # Lower Bounds (Amps > 0, Sigma > 0)
    lb = [0, b1 - delta_b1, 0, 0, b2 - delta_b2, 0]
    # Upper Bounds
    ub = [np.inf, b1 + delta_b1, np.inf, np.inf, b2 + delta_b2, np.inf]

    try:
        popt, pcov = curve_fit(
            double_gaussian, xdata, ydata, p0=fitparams, bounds=(lb, ub), maxfev=5000
        )
    except Exception as e:
        print(f"Double gauss fit failed: {e}")
        popt = fitparams  # Return guess if fail
        pcov = None

    return popt, pcov


# --- T1 Decay Physics Models ---


def distfn(v, vg, ve, sigma, tm):
    """Distribution function modeling T1 decay."""
    dv = ve - vg
    if dv == 0:
        dv = 1e-9

    term1 = tm * sigma**2 / (2 * dv**2)
    term2 = (v - vg) / dv

    arg1 = (tm * sigma**2 / dv + ve - v) / (np.sqrt(2) * sigma)
    arg2 = (-tm * sigma**2 / dv + v - vg) / (np.sqrt(2) * sigma)

    return np.abs(
        (tm / (2 * dv)) * np.exp(tm * (term1 - term2)) * (erf(arg1) + erf(arg2))
    )


def excited_func(x, vg, ve, sigma, tm):
    """Fit function for excited state including T1 decay."""
    return gaussian(x, 1, ve, sigma) * np.exp(-tm) + distfn(x, vg, ve, sigma, tm)


def fit_gaussian_data_wrapper(d, nbins=100, p0=None):
    """Helper: Takes raw data, makes hist, fits single gaussian."""
    v, hist_val = make_hist(d, nbins)
    if p0 is None:
        p0 = [np.mean(v * hist_val) / np.mean(hist_val), np.std(d)]
        # Add amplitude estimate (area under curve approx 1)
        p0 = [1, p0[0], p0[1]]

    params = fit_gauss_helper(v, hist_val, p0)
    return params, v, hist_val


# ===================================================================== #
# Section 2: Visualization (General Hist)
# ===================================================================== #
# (保留你原本的 general_hist, 這裡省略以節省篇幅，確保 fit_single_shot 能跑即可)
# 若需要完整代碼請將前一段的 general_hist 貼回此處。

# ===================================================================== #
# Section 3: Advanced Analysis (Updated fit_single_shot)
# ===================================================================== #


def fit_single_shot(d, plot=True, rot=True, double_fit=True):
    """
    Comprehensive analysis including Single Gauss, Double Gauss (Thermal Pop), and T1 Decay.
    """
    d = deepcopy(d)

    # 1. Rotation
    if rot:
        xg, yg = np.mean(d["Ig"]), np.mean(d["Qg"])
        xe, ye = np.mean(d["Ie"]), np.mean(d["Qe"])
        theta = -np.arctan2((ye - yg), (xe - xg))
        data = full_rotate(d, theta)
    else:
        data = d
        theta = 0

    # 2. Basic Single Gaussian Fit (Establish Baseline)
    # Params structure: [mag, center, sigma]
    paramsg, vhg, histg = fit_gaussian_data_wrapper(data["Ig"], nbins=100)
    paramsqg, vhqg, histqg = fit_gaussian_data_wrapper(data["Qg"], nbins=100)
    paramsqe, vhqe, histqe = fit_gaussian_data_wrapper(data["Qe"], nbins=100)

    vg = paramsg[1]  # I center for ground
    sigma = paramsg[2]  # Width (from ground state)
    vqg = paramsqg[1]
    vqe = paramsqe[1]

    # 3. T1 Decay Fit on Excited State
    vhe, histe = make_hist(data["Ie"], nbins=100)
    p0_t1 = [np.mean(vhe * histe) / np.mean(histe), 0.2]  # [center, tm]

    try:
        paramse_t1, _ = curve_fit(
            lambda x, ve, tm: excited_func(x, vg, ve, sigma, tm), vhe, histe, p0=p0_t1
        )
        ve = paramse_t1[0]
        tm = paramse_t1[1]
    except:
        ve = p0_t1[0]
        tm = 0
        print("T1 Fit Failed")

    # 4. Double Gaussian Fit on Ground State (Check Thermal Population)
    # We suspect the 'g' histogram might contain a small 'e' peak
    fit_results_2g = None
    thermal_pop = 0.0

    if double_fit:
        # Initial Guess:
        # Peak 1 (Main G): Amp ~0.95, Center=vg, Width=sigma
        # Peak 2 (Small E): Amp ~0.05, Center=ve, Width=sigma
        guess_2g = [
            0.95,
            vg,
            sigma,  # G Component
            0.05,
            ve,
            sigma,  # E Component (Leakage)
        ]

        try:
            popt_2g, _ = fit_doublegauss_helper(vhg, histg, guess_2g)

            # Calculate thermal population ratio based on Amplitudes
            # Area ~ Amplitude (since width is similar). Ideally Area = Amp * Width * sqrt(2pi)
            # Assuming widths are similar, ratio of Amps is approx ratio of population
            amp_g = popt_2g[0]
            amp_e = popt_2g[3]
            thermal_pop = amp_e / (amp_g + amp_e)

            fit_results_2g = popt_2g
        except Exception as e:
            print(f"Double Gauss Fit Logic Error: {e}")

    # 5. Plotting
    if plot:
        plt.figure(figsize=(10, 5))

        # Plot Data
        plt.plot(vhg, histg, "b.", alpha=0.2, label="Data |g>")
        plt.plot(vhe, histe, "r.", alpha=0.2, label="Data |e>")

        x_dense = np.linspace(min(vhg[0], vhe[0]), max(vhg[-1], vhe[-1]), 500)

        # Plot Single Fits
        plt.plot(
            x_dense,
            gaussian(x_dense, 1, vg, sigma),
            "b--",
            lw=1,
            alpha=0.5,
            label="Single Gauss |g>",
        )
        plt.plot(
            x_dense,
            excited_func(x_dense, vg, ve, sigma, tm),
            "r-",
            lw=2,
            label=f"T1 Fit |e> (Tm={tm:.2f})",
        )

        # Plot Double Fit if successful
        if fit_results_2g is not None:
            y_2g = double_gaussian(x_dense, *fit_results_2g)
            # Plot the composite fit
            plt.plot(
                x_dense,
                y_2g,
                "k-",
                lw=2,
                label=f"Double Gauss |g> (Therm={thermal_pop * 100:.1f}%)",
            )
            # Plot the leakage component separately for visibility
            y_leakage = gaussian(
                x_dense, fit_results_2g[3], fit_results_2g[4], fit_results_2g[5]
            )
            plt.fill_between(
                x_dense, y_leakage, color="orange", alpha=0.3, label="Thermal Pop. Fit"
            )

        plt.ylabel("Probability Density")
        plt.xlabel("I [Rotated]")
        plt.title(f"Readout Analysis: T1 Decay & Thermal Population")
        plt.legend()
        plt.tight_layout()
        plt.show()

    # 6. Return Data
    fit_params = {
        "theta": theta,
        "vg": vg,
        "ve": ve,
        "sigma": sigma,
        "tm": tm,
        "thermal_pop": thermal_pop,
        "double_gauss_params": fit_results_2g,
    }

    return data, fit_params
```
